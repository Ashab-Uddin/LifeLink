```python
import pandas as pd
import torch
from torch.utils.data import DataLoader, TensorDataset
import numpy as np

# Seed once globally
np.random.seed(42)

# Step 1: Custom Stratified Split (77 train, 19 val, 24 test per class)
train_indices, val_indices, test_indices = [], [], []
for class_id in range(len(label_encoder.classes_)):  # total 41 classes
    # Get indices for this class
    class_indices = np.where(y == class_id)[0]
    
    # Shuffle only once per class
    np.random.shuffle(class_indices)
    
    train_indices.extend(class_indices[:77])
    val_indices.extend(class_indices[77:96])    # 19 val
    test_indices.extend(class_indices[96:120])  # 24 test

# Step 2: Create tensors for each split
X_train = X[train_indices]
y_train = y[train_indices]
X_val = X[val_indices]
y_val = y[val_indices]
X_test = X[test_indices]
y_test = y[test_indices]

# Step 3: Print to verify class balance
print("✅ Class distribution in y_train:", torch.bincount(y_train))
print("✅ Class distribution in y_val:", torch.bincount(y_val))
print("✅ Class distribution in y_test:", torch.bincount(y_test))

# Step 4: Create DataLoaders
train_dataset = TensorDataset(X_train, y_train)
val_dataset = TensorDataset(X_val, y_val)
test_dataset = TensorDataset(X_test, y_test)

train_dl = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_dl = DataLoader(val_dataset, batch_size=32, shuffle=False)
test_dl = DataLoader(test_dataset, batch_size=32, shuffle=False)

# Step 5: Shapes check
print("📦 X_train shape:", X_train.shape)
print("📦 y_train shape:", y_train.shape)
print("📦 X_val shape:", X_val.shape)
print("📦 y_val shape:", y_val.shape)
print("📦 X_test shape:", X_test.shape)
print("📦 y_test shape:", y_test.shape)
```

```python
model = nn.Sequential(

    nn.Linear(132, 82).double(),
    nn.ReLU(),
    nn.Dropout(0.2),

    nn.Linear(82, 41).double(),
    nn.Softmax(dim=-1)
)
model
```

```python
# create loss function
loss_fn = nn.CrossEntropyLoss()

# create optimizer
optimizer = optim.SGD(model.parameters(), lr=0.001, momentum=0.95)
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

```python
def train_one_epoch():
    model.train()  # set the model to training mode
    train_loss = 0.0
    correct = 0  # to track correct predictions
    total = 0    # to track total samples

    for features, targets in train_dl:
        # 1. Set the gradients to zero
        optimizer.zero_grad()
        
        # 2. Forward pass
        preds = model(features)
        
        # 3. Compute loss
        loss = loss_fn(preds, targets)
        
        # 4. Backward pass and optimize
        loss.backward()
        optimizer.step()
        
        # 5. Update training loss
        train_loss += loss.item()
        
        # 6. Calculate accuracy for this batch
        _, predicted = torch.max(preds, 1)  # Get the index of the max log-probability
        total += targets.size(0)            # Total number of labels
        correct += (predicted == targets).sum().item()  # Number of correct predictions

    # Calculate average training loss and accuracy
    avg_train_loss = train_loss / len(train_dl)
    train_accuracy = correct / total  # Average accuracy over all batches

    return avg_train_loss, train_accuracy  # Return both loss and accuracy


def validate_with_accuracy():
    model.eval()  # set the model to evaluation mode
    val_loss = 0.0
    correct = 0
    total = 0
    with torch.no_grad():  # no need to compute gradients during validation
        for X, y in val_dl:
            preds = model(X)
            loss = loss_fn(preds, y)
            val_loss += loss.item()
            
            _, predicted = torch.max(preds, 1)
            total += y.size(0)
            correct += (predicted == y).sum().item()
            
    avg_val_loss = val_loss / len(val_dl)        
    accuracy = correct / total
    return avg_val_loss, accuracy


def train_model(num_epochs):
    train_losses = []
    train_accuracies = []
    val_losses = []
    val_accuracies = []

    best_val_acc = 0.0  # 🌟 Track best validation accuracy
    best_model_state = None  # To store best weights

    for epoch in range(num_epochs):
        train_loss, train_accuracy = train_one_epoch()
        train_losses.append(train_loss)
        train_accuracies.append(train_accuracy)

        avg_val_loss, val_accuracy = validate_with_accuracy()
        val_losses.append(avg_val_loss)
        val_accuracies.append(val_accuracy)

        
        # ✅ Save best model only if current val acc is higher
        if val_accuracy > best_val_acc:
            best_val_acc = val_accuracy
            best_model_state = model.state_dict()

        # Print progress
        if ((epoch + 1) % 10) == 0:
            print(f'Epoch {epoch+1}/{num_epochs}, Train Loss: {train_loss:.4f}, Train Acc: {train_accuracy:.4f}, Validation Loss: {avg_val_loss:.4f}, Validation Acc: {val_accuracy:.4f}')

    # 🧠 Restore the best model after training
    if best_model_state is not None:
        model.load_state_dict(best_model_state)
        print(f"\n🫶 Best model restored with validation accuracy: {best_val_acc:.4f}")
    
    torch.save(model.state_dict(), 'pytorch_model.pt')


    return train_losses, train_accuracies, val_losses, val_accuracies


def plot_losses_and_accuracy(train_losses, train_accuracies, val_losses, val_accuracies):
    plt.figure(figsize=(12, 8))
    plt.plot(train_losses, label='Training Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.title('Training and Validation Losses')
    plt.legend()
    plt.show()

    plt.figure(figsize=(12, 8))
    plt.plot(train_accuracies, label='Training Accuracy')
    plt.plot(val_accuracies, label='Validation Accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy')
    plt.ylim(0, 1)  # Set y-axis limit to 0-1 for accuracy
    plt.title('Training and Validation Accuracies')
    plt.legend()
    plt.savefig('accuracy_plot.png')
    plt.show()
```

```python
train_losses, train_accuracies, val_losses, val_accuracies = train_model(num_epochs=600)
plot_losses_and_accuracy(train_losses, train_accuracies, val_losses, val_accuracies)
pytorch_acc = val_accuracies[-1]
pytorch_acc
```

```python
def test_accuracy_and_stats(label_encoder):
    model.eval()  # Set model to evaluation mode
    all_preds = []
    all_true = []
    with torch.no_grad():
        for X, y in test_dl:
            preds = model(X)
            _, predicted = torch.max(preds, 1)
            all_preds.extend(predicted.cpu().numpy())
            all_true.extend(y.cpu().numpy())

    # Compute overall accuracy
    overall_accuracy = accuracy_score(all_true, all_preds)

    # Initialize counters for each class
    class_counts = {label: 0 for label in label_encoder.classes_}
    correct_counts = {label: 0 for label in label_encoder.classes_}

    # Count correct predictions and total instances for each class
    for true, pred in zip(all_true, all_preds):
        true_label = label_encoder.inverse_transform([true])[0]
        pred_label = label_encoder.inverse_transform([pred])[0]
        class_counts[true_label] += 1
        if true == pred:
            correct_counts[true_label] += 1

    # Prepare data for DataFrame
    data = []
    for label in label_encoder.classes_:
        accuracy = correct_counts[label] / class_counts[label] if class_counts[label] > 0 else 0
        data.append({
            'Class': label,
            'Total Predictions': class_counts[label],
            'Correct Predictions': correct_counts[label],
            'Accuracy': accuracy
        })

    # Convert data to DataFrame
    df_stats = pd.DataFrame(data)

    return overall_accuracy, df_stats
```

```python
overall_accuracy, df_stats = test_accuracy_and_stats(label_encoder)
print(f"Overall Test Accuracy: {overall_accuracy:.4f}")
df_stats
```

```python
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

def evaluate_and_report(data_loader, class_names):
    model.eval()
    y_true = []
    y_pred = []

    with torch.no_grad():
        for X, y in data_loader:
            preds = model(X)
            _, predicted = torch.max(preds, 1)
            
            y_true.extend(y.tolist())           # actual labels
            y_pred.extend(predicted.tolist())   # predicted labels
    
    # 💬 Classification Report
    print("📝 Classification Report:\n")
    print(classification_report(y_true, y_pred, target_names=label_encoder.classes_))

    # 📊 Confusion Matrix
    print("📊 Confusion Matrix:")
    cm = confusion_matrix(y_true, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=label_encoder.classes_)
    fig, ax = plt.subplots(figsize=(18, 12))
    disp.plot(xticks_rotation=90, cmap='Greens', values_format='d', ax = ax)
    ax.tick_params(axis='x', labelsize=12, rotation=90)
    ax.tick_params(axis='y', labelsize=12)
    plt.title("Confusion Matrix", fontsize=16)
    plt.show()

    return y_true, y_pred  # in case you want to use them later

# Example: assuming you have your class names list
class_names = [
    "(vertigo) Paroymsal Positional Vertigo", "AIDS", "Acne", ..., "hepatitis A"
]

y_true, y_pred = evaluate_and_report(test_dl, class_names)

# Decode if needed
y_true = label_encoder.inverse_transform(y_true)
y_pred = label_encoder.inverse_transform(y_pred)

print(classification_report(y_true, y_pred))
```
